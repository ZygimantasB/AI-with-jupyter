```python
import os

from dotenv import load_dotenv
from lxml.html import document_fromstring
from mistralai import Mistral
from nbconvert.filters.strings import files_url_pattern
```

```python
load_dotenv()

client = Mistral(api_key=os.getenv("MISTRAL_API_KEY"))

pdf_file_path = r"C:\Users\Zygim\Downloads\file-example_PDF_500_kB.pdf"
```

```python
upload_file = client.files.upload(
    file = {
        'file_name': "attention.pdf",
        'content': open(pdf_file_path, "rb")
    },
    purpose='ocr'
)
```

```python
file_url = client.files.get_signed_url(file_id=upload_file.id)
```

```python
response = client.ocr.process(
    model="mistral-ocr-latest",
    document={
        'type': "document_url",
        "document_url": file_url.url,
    },
    include_image_base64=True
)
```

```python
response.pages[0].markdown
```

```python
import base64
import os

output_dir = 'mistral_images'
os.makedirs(output_dir, exist_ok=True)

def export_image(image):
    header, encoded = image.image_base64.split(",", 1)

    mime_type = header.split(';')[0].split(':')[1]
    extension = mime_type.split('/')[1]

    parsed_image = base64.b64decode(encoded)

    file_path = os.path.join(output_dir, f"{image.id}.{extension}")

    with open(file_path, "wb") as file:
        file.write(parsed_image)

output_md_path = os.path.join(output_dir, 'output.md')

with open(output_md_path, 'w', encoding='utf-8') as file_url:
    for page in response.pages:
        file_url.write(page.markdown)
        for image in page.images:
            export_image(image)

print(f"Markdown and images have been saved to the '{output_dir}' directory.")
```

```python
from pdf2image import convert_from_bytes
import pytesseract
import os

poppler_bin_path = r"F:\DataSpell\AI-with-jupyter\YouTube\Fine Tunning The Model\poppler-24.08.0\Library\bin"

try:
    with open(pdf_file_path, "rb") as file:
        pdf_bytes = file.read()

    images = convert_from_bytes(pdf_bytes, poppler_path=poppler_bin_path)

    custom_config = r'--oem 3 --psm 6'

    pdf_text = '\n\n'.join(
        pytesseract.image_to_string(image, config=custom_config)
        for image in images
    )

    print(pdf_text)

except Exception as e:
    print(f"An error occurred: {e}")
```
