```python
import fitz  # PyMuPDF
import easyocr
import os

PDF_PATH = r"C:\Users\Zygim\Downloads\file-example_PDF_500_kB.pdf"
LANGUAGES = ['en']
DPI = 300

def extract_text_from_pdf(pdf_path):
    if not os.path.exists(pdf_path):
        return f"Error: The file '{pdf_path}' was not found."

    print("Initializing EasyOCR... This might take a moment.")
    reader = easyocr.Reader(LANGUAGES, gpu=True)

    print(f"Opening PDF: {pdf_path}")
    doc = fitz.open(pdf_path)

    full_text = []

    for page_num, page in enumerate(doc):
        print(f"Processing page {page_num + 1} of {len(doc)}...")

        pix = page.get_pixmap(dpi=DPI)
        image_bytes = pix.tobytes("png")

        result = reader.readtext(image_bytes, detail=0, paragraph=True)

        full_text.extend(result)

    doc.close()

    return "\n".join(full_text)


if __name__ == '__main__':
    extracted_text = extract_text_from_pdf(PDF_PATH)

    print("\n--- EXTRACTED TEXT ---")
    print(extracted_text)

    with open("extracted_output.txt", "w", encoding="utf-8") as f:
        f.write(extracted_text)

    print("\n--- PROCESSING COMPLETE ---")
    print("Extracted text has been saved to 'extracted_output.txt'")
```
